## All groups info

```python
with open("../output/groups_info.json", "r") as f:
    groups_info = json.load(f)
```

```python
df = pd.DataFrame(groups_info).T

df['unique_tags_pr_work'] = df['n_unique_tags'] / df['n_works']
```

## Harry Potter

```python
with open("../output/Harry Potter/slash_tags_counter.json", "r") as f:
    hp_slash = json.load(f)
```

```python
hp_slash_df = pd.DataFrame.from_dict(hp_slash,orient="index" ).reset_index()
hp_slash_df = hp_slash_df.rename(columns={'index':'tag', 0:'count'})

hp_slash_df.sort_values("count", ascending=False).head(50)
```
