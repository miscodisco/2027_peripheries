## All groups info

```python
with open("../output/groups_info.json", "r") as f:
    groups_info = json.load(f)
```

```python
df = pd.DataFrame(groups_info).T

df['unique_tags_pr_work'] = df['n_unique_tags'] / df['n_works']
```

```python
with open("../output/all_tags.json", "r") as f:
    all_tags = json.load(f)
```

```python
tags_df = pd.DataFrame.from_dict(all_tags,orient="index" ).reset_index()
tags_df = tags_df.rename(columns={'index':'tag', 0:'count'})

tags_df.sort_values("count", ascending=False).head(50)
```

```python
total_tag_count = tags_df['count'].sum()
total_tag_count
```

```python
tags_df['freq'] = (tags_df['count'] / total_tag_count)*100
```

```python
tags_df
```

```python
tags_less_five = tags_df.loc[tags_df['count'] < 5]

tags_less_five
```

```python
tags_subset = tags_df.loc[tags_df['count'] >= 5]
```
